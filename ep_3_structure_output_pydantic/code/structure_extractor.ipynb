```python

import os
from pathlib import Path  ## used to read .txt file
from dotenv import load_dotenv ## used to read .env file
from pydantic import BaseModel,Field
from langchain.chat_models import init_chat_model
```

```python
load_dotenv()
```

```python
MODEL_NAME = "gemini-3.5-flash"
print("gemini model:",MODEL_NAME)
```

```python
class Resume(BaseModel):
    name : str=Field(description="full name of the candidate")
    email: str | None =Field(description="email if present ,else none")
    years_experiance: float | None =Field(description="year of experience if present, else none")
    skills : list[str] =Field(description="list of skill mentioned ")
```

```python
data_dir=Path("data")
text=(data_dir/"resume_data.txt").read_text(encoding="utf-8")
```

```python
llm=init_chat_model(MODEL_NAME,model_provider="google_genai", temperature=0)
```

```python
prompt=f"from given resume {text} /n extract name, email ,year of experience and skills in json formate."

responce=llm.invoke(prompt)
print(responce)

```
